## 7. Process Demographics — replace item numbers with labels

```python
# Load demographics and constructs
demo = pd.read_csv(DATA_DIR / 'Demographics.csv')
constructs = pd.read_excel(DATA_DIR / 'Constructs items.xlsx', sheet_name=None)

# Build Q → Original_Item and Q → Label lookups per sheet
label_lookups = {
    sheet: dict(zip(df['Q'], df['Label']))
    for sheet, df in constructs.items()
}
original_lookups = {
    sheet: dict(zip(df['Q'], df['Original_Item']))
    for sheet, df in constructs.items()
}

# Map each demographics column to its constructs sheet (None = blank out)
# Note: Triggers_items already contains text labels, not Q numbers — kept as-is
col_to_sheet = {
    'Appraisals_items':      'Appraisals',
    'Memory_items':          'Memory nature',
    'Threat_items_original': 'Threat',
    'Threat_items':          'Threat',
    'Strategies_items':      'Strategies',
    'Avoidance_items':       'Avoidance strategies',
    'RIQ_items':             None,
    'WEMWBS_items':          None,
}

SEP = ' | '  # separator safe from commas inside item text

def replace_with_values(value, lookup):
    if pd.isna(value):
        return value
    qs = [int(float(q.strip())) for q in str(value).split(',') if q.strip()]
    return SEP.join(lookup.get(q, str(q)) for q in qs)

demo_out = demo.copy()
for col, sheet in col_to_sheet.items():
    if sheet is None:
        demo_out[col] = None
    else:
        idx = demo_out.columns.get_loc(col)
        demo_out.insert(idx + 1, col + '_label',    demo_out[col].apply(lambda v, s=sheet: replace_with_values(v, label_lookups[s])))
        demo_out.insert(idx + 2, col + '_original', demo_out[col].apply(lambda v, s=sheet: replace_with_values(v, original_lookups[s])))
        demo_out = demo_out.drop(columns=[col])

# Drop unnamed columns
demo_out = demo_out.loc[:, ~demo_out.columns.str.startswith('Unnamed')]

demo_out.head(3)
```

```python
demo_out.to_csv(OUTPUT_DIR / 'demographics_labeled.csv', index=False)
print(f'Saved: demographics_labeled.csv  ({len(demo_out)} rows, {len(demo_out.columns)} cols)')
```
