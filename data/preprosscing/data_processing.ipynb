# PTSD Simulation — Data Processing

Loads and processes the two source Excel files:
- `MSN data processed.xlsx` — per-patient per-timepoint node/edge scores
- `MSN edge aggregated data.xlsx` — per-patient aggregated edge statistics

Exports clean CSVs for use in the simulation pipeline.

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path('.')
OUTPUT_DIR = DATA_DIR
```

## 1. Load raw data

```python
df_msn = pd.read_excel(DATA_DIR / 'MSN data processed.xlsx')
df_edge = pd.read_excel(DATA_DIR / 'MSN edge aggregated data.xlsx')

print('MSN data processed:  ', df_msn.shape)
print('MSN edge aggregated: ', df_edge.shape)
```

## 4. Process MSN data processed

Edit this section to add any cleaning or transformations you need.

```python
df_msn_clean = df_msn.copy()

df_msn_clean = df_msn_clean.drop(columns=['Subject', 'Date', "Time", "Day"])
df_msn_clean = df_msn_clean.rename(columns={"Subnum":"id", "Date.and.time" : "datetime"})

df_msn_clean.head()
```

## 5. Process MSN edge aggregated data

Edit this section to add any cleaning or transformations you need.

```python
df_edge_clean = df_edge.copy()

df_edge_clean = df_edge_clean.rename(columns={"Subnum" : "id"})
df_edge_clean = df_edge_clean.drop(columns={"Edge", "statistic", "Condition", "CI_lower", "CI_upper"})

df_edge_clean.head()
```

## 6. Export to CSV

```python
msn_out = OUTPUT_DIR / 'ema_data.csv'
edge_out = OUTPUT_DIR / 'aggregated_ema_data.csv'

df_msn_clean.to_csv(msn_out, index=False)
df_edge_clean.to_csv(edge_out, index=False)

print(f'Saved: {msn_out}  ({len(df_msn_clean)} rows)')
print(f'Saved: {edge_out}  ({len(df_edge_clean)} rows)')
```
